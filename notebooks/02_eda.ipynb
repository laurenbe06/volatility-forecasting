```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from statsmodels.tsa.stattools import adfuller
from statsmodels.stats.diagnostic import acorr_ljungbox, het_arch
from statsmodels.graphics.tsaplots import plot_acf

df = pd.read_csv("../data/GSPC_processed.csv", index_col=0, parse_dates=True)
r = df["log_return"]
```

```python
adf = adfuller(r)
print(f"ADF stat: {adf[0]:.3f}, p-value: {adf[1]:.4g}")
```

```python
print("Skew:", stats.skew(r), " Excess kurtosis:", stats.kurtosis(r))
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(r, bins=100, density=True, alpha=0.7)
x = np.linspace(r.min(), r.max(), 300)
ax[0].plot(x, stats.norm.pdf(x, r.mean(), r.std()), "r")
ax[0].set_title("Returns vs normal")
stats.probplot(r, plot=ax[1]); plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
plot_acf(r, lags=40, ax=ax[0], title="ACF of returns")
plot_acf(r**2, lags=40, ax=ax[1], title="ACF of squared returns")
plt.tight_layout(); plt.show()
```

```python
print(acorr_ljungbox(r**2, lags=[10, 20], return_df=True))
lm, lm_p, f, f_p = het_arch(r, nlags=10)
print(f"\nEngle ARCH LM test: stat={lm:.2f}, p={lm_p:.4g}")
```
