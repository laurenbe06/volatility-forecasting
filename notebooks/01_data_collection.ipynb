```python
import sys
sys.path.append("..")

import matplotlib.pyplot as plt
from src.utils import download_prices, add_returns, realized_vol

TICKER = "^GSPC"
prices = download_prices(TICKER, start="2010-01-01")
prices.head()
```

```python
df = add_returns(prices)
df["realized_vol_21d"] = realized_vol(df["log_return"], window=21)
df.describe()
```

```python
fig, ax = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
df["log_return"].plot(ax=ax[0], lw=0.6, title="Daily log returns (%)")
df["realized_vol_21d"].plot(ax=ax[1], title="21-day realized volatility (annualized, %)")
plt.tight_layout(); plt.show()
```

```python
df.to_csv("../data/GSPC_processed.csv")
print(df.shape, df.index.min().date(), "->", df.index.max().date())
```
